```python
with open('data/BERT.pdf','rb')as fp:   #rb is read binary mode
    reader=PyPDF2.PdfReader(fp)
    pages=[]
    for i in range(len(reader.pages)):
        page=reader.pages[i]
        text=page.extract_text()
        pages.append(text)
```

```python
pages
```

```python
text="\n".join(pages)
with open('data/bert.txt','w')as fp:
    fp.write(text)
```

```python
pages=[]
with open('data/BERT.pdf','rb')as fp:   #rb is read binary mode
    reader=PyPDF2.PdfReader(fp)
    
    for i in range(len(reader.pages)):
        if i+1 in[1,2,3]:
            page=reader.pages[i]
            text=page.extract_text()
            pages.append(text)

text="\n".join(pages)

with open('data/bert123.txt','w')as fp:
    fp.write(text)        
```
