## Read and Evaluate the Files
- Reading Files: How to read the files and evaluate the content.
- Evaluating Expressions from Files: Using eval() to interpret data.

```python
#There are various ways to read the content in the file but various issues tag along with these ways.Now we will see each step of it
fp=open('data/example.txt','r')
fp.read()
```

```python
fp=open('data/example.txt','r')
Text=fp.read()
print(Text)
Text.splitlines() #Gives us the separate lines enclosed in single inverted commas
```

```python
fp=open('data/example.txt','r')
Text=fp.readlines() # Reads all the lines  #\n will be seen
Text
```

```python
fp=open('data/example.txt','r')
Text=fp.readline() # Just reads the single line  #\n will be seen
Text
```

```python
Text=fp.readline() # Now the file pointer is at the end of first line so next lin will be printed
Text
fp.close()
```
