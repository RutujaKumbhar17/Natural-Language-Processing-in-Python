## Working with open() Files in write() Mode
- Opening and Writing to Files: How to create and write to text files.
- Appending to Files: How to append content to an existing file.

```python
import os

if not os.path.exists('Data'): #creats'data'if it doesnt exist, necessary to check whether it ecists because if it already exists, it will throw error
    os.makedirs('data')

os.makedirs('data',exist_ok=True)    
```

```python
fp=open('data/example.txt','w') #fp-file path, name of the file must be unique or else it will over write, 'w' is for write mode.
fp.write('This content is written through my code.\n') #\n takes the pointer to the next line.
fp.write('This is the next line, to bring this content to the next line, either write new line  in the end to previous line or at hte begining of this line')
fp.close() #It's very important to close the file as if you don't do so the content will be written in the cache and won't be visible in the actual file.
```

```python
with open('data/example1.txt','w')as fp:  #this kinda code looks elegant  #The file is created by itself
    fp.write('Hello\n')
```
