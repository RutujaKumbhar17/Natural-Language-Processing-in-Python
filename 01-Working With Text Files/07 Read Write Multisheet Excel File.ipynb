## Reading and Writing .XLSX Files with Pandas
- Working with Exce Files: How to read and write Excel Files using Pandas.
- Multiple Sheets:Handle Excel Files with multiple sheets. 

```python
import pandas as pd
url = 'https://github.com/laxmimerit/All-CSV-ML-DATA-Files-Download/raw/master/Data%20-%20Multiple%20Worksheets.xlsx'
df = pd.read_excel(url)
df
```

```python
#Check the number of sheets in the file

df=pd.read_excel(url,sheet_name=None)
df
# It has two sheets,if sheet name not provided it takes first sheet by default
```

```python
df=pd.read_excel(url)  # sheet name not given so it took first sheet
df
```

```python
df=pd.read_excel(url, sheet_name=None) #Here it is necessary to give sheet_name =None as it needs to acces all sheets
df.items()
df1=df['Data 1']
df2=df['Data 2']
```

```python
df2
```

```python
df=pd.read_excel(url, sheet_name=None, usecols=['First Name'])
df
```

```python
df1.to_excel('data/df1.xlsx',index=None)
```

```python
with pd.ExcelWriter('data/multiple_sheet.xlsx')as fp: #we can writ in the excwl file
    df1.to_excel(fp,sheet_name='Data 1')
    df2.to_excel(fp,sheet_name='Data 2')
```
