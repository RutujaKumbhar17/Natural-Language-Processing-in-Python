```python
lines=['first line','second line','third line']

for line in lines:
    with open('data/example2.txt','a')as fp:  #'a' is for append mode
        fp.write(line)
```

```python
#To solve the problem of new line either add new line after all the lines or include it inn the for loop.
lines=['first line','second line','third line']
# lines=['first line\n','second line\n','third line\n']

for line in lines:
    with open('data/example2.txt','w')as fp: 
        fp.write(line)
        fp.write('\n')
```

```python
#lines=['first line','second line','third line']
lines=['first line\n','second line\n','third line\n']

with open('data/example2.txt','w')as fp:  
   #To do it in one line
   fp.writelines(lines) #to use this we need to add the \n at the end of each line
```
