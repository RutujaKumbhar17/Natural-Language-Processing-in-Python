### 6. **Chunking**
Chunking groups words into meaningful chunks, like noun phrases.


```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer
from nltk import pos_tag
from prettytable import PrettyTable
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
from nltk.chunk import RegexpParser
text = " ".join(open('data/bert_extracted_text.txt').read().splitlines())
grammar = "NP: {<DT>?<JJ>*<NN>}"
```

```python
text = 'The scientists discover new species every year, Last year, they discovered an ancient artifact. They are discovering new techniques with their recent discovery'
stemmer = PorterStemmer()
words = word_tokenize(text)
words
```

```python
tagged_words = pos_tag(words)
```

```python
#pip install svgling
chunk_parser = RegexpParser(grammar)
chunks = chunk_parser.parse(tagged_words)
```
