### 2. **Filtering Stop Words**
Stop words are common words that are usually removed in text processing tasks.


```python
from nltk.corpus import stopwords
import spacy
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize
```

```python
nltk.download('stopwords')
```

```python
stop_words = stopwords.words('english')
```

```python
stop_words
```

```python
nlp = spacy.load('en_core_web_sm')
spacy_stop_words = nlp.Defaults.stop_words
```

```python
stop_words
```

```python
len(stop_words),len(spacy_stop_words)
```

```python
text = " ".join(open('data/bert_extracted_text.txt').read().splitlines())
```

```python
words = word_tokenize(text.lower())
```

```python
sw_len = set([word for word in words if word in stop_words])
len(sw_len)
```

```python
print(' '.join([word for word in words if word not in stop_words]))
```
