### 4. **Tagging Parts of Speech (POS Tagging)**
POS tagging assigns grammatical tags to each token, such as noun, verb, etc.


```python
import nltk
from nltk import pos_tag
from nltk.stem import PorterStemmer
from nltk.tokenize import word_tokenize
from prettytable import PrettyTable
# These are standard dependencies for these functions
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
text = 'The scientists discover new species every year, Last year, they discovered an ancient artifact. They are discovering new techniques with their recent discovery'
stemmer = PorterStemmer()
words = word_tokenize(text)
words
```

```python
nltk.download('averaged_perceptron_tagger')
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger_eng') # This fixes the POS Tagging error
```

```python
tagged_words = pos_tag(words)
```

```python
tagged_words
```

```python
nltk.download('tagsets') 
# This fixes the help.upenn_tagset() error
nltk.download('tagsets_json')
nltk.help.upenn_tagset()
```

```python
table = PrettyTable(field_names=['word','pos tag'])
for word, pos in zip(words, tagged_words):
    table.add_row([word, pos[1]])

print(table)
```
