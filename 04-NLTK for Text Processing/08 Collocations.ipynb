### 8. **Finding Collocations**
Collocations are word pairs that frequently appear together.

```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer
from nltk import pos_tag
from prettytable import PrettyTable
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
text = " ".join(open('data/bert_extracted_text.txt').read().splitlines())
```

```python
text
```

```python
from nltk.collocations import BigramCollocationFinder
from nltk.metrics import BigramAssocMeasures
```

```python
bigram = BigramCollocationFinder.from_words(word_tokenize(text))
```

```python
bigram
```

```python
scores = bigram.nbest(BigramAssocMeasures.likelihood_ratio,10)
```
