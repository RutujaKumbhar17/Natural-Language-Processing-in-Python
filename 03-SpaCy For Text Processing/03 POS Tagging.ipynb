## 2.Parts Of Speech (POS) Tagging

```python
import spacy
import prettytable
```

```python
nlp = spacy.load('en_core_web_sm')
text="Hello I'm learning NLP.This is the SpaCy lesson."
doc=nlp(text)
```

```python
doc
```

```python
table=prettytable.PrettyTable()
table.field_names=['token','pos','details','explanation']

for token in doc:
    table.add_row([token.text,token.pos_,token.tag_,spacy.explain(token.tag_)])

print(table)
```
