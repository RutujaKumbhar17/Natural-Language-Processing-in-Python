## SpaCy
- spaCy takes "Text" and returns a "Doc" object. > That "Doc" object isn't just a string anymore; it's a bundle of linguistic intelligence where every word knows its own grammar, its base form, and its relationship to its neighbors

- spaCy is a free, open-source library for advanced Natural Language Processing (NLP) in Python.

- spaCy is designed specifically for production use and helps you build applications that process and “understand” large volumes of text.

- It can be used to build information extraction or natural language understanding systems, or to pre-process text for deep learning.

### Point-by-point breakdown of what spaCy does:

- Tokenization: Breaks raw text into individual pieces (words, prefixes, suffixes, and punctuation).

- POS Tagging: Assigns grammatical labels to each word (e.g., Noun, Verb, Adjective).

- Lemmatization: Reduces words to their "dictionary" or base form (e.g., "running" → "run").

- Dependency Parsing: Maps the grammatical structure to show how words relate to each other (who did what to whom).

- Named Entity Recognition (NER): Identifies and categorizes "real-world" objects like Names, Dates, Organizations, and Locations.

- Sentence Segmentation: Automatically detects where one sentence ends and the next begins.

- Attribute Scoring: Checks if a word is a "stop word" (like 'the' or 'is'), a number, or an email address.

- Word Vectors: Converts words into mathematical coordinates so the computer can measure how similar two words are in meaning.
