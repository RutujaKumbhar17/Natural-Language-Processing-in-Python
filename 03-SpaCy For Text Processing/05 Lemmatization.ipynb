### 4. **Lemmatization**


1. **Base Form Reduction**: Lemmatization reduces words to their meaningful base form (e.g., "running" to "run").

2. **Enhances NLP Accuracy**: It normalizes word variations, improving the accuracy of NLP tasks by treating different word forms as the same.

```python
import spacy
import prettytable
```

```python
nlp = spacy.load('en_core_web_sm')
```

```python
text='dance dancing danced'
doc=nlp(text)
table=prettytable.PrettyTable(field_names=['token','lemma'])
for token in doc:
    table.add_row([token.text,token.lemma_])
print(table)

#In spaCy, the lemmatizer relies on the context (the POS tag) to decide the base form.

#If "dancing" is identified as a Noun (a gerund, like "Dancing is fun"), spaCy often keeps it as "dancing."

#If "dancing" is identified as a Verb (like "They are dancing"), spaCy will correctly change it to "dance."
```

```python
text = 'The scientists discover new species every year, Last year, they discovered an ancient artifact. They are discovering new techniques with their recent discovery'

doc = nlp(text)

table = prettytable.PrettyTable(field_names=['token', 'lemma', 'pos'])
for token in doc:
    table.add_row([token.text, token.lemma_, token.pos_])

print(table)
```
