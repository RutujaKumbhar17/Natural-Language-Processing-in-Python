## 1.Tokenization

```python
import spacy
from prettytable import PrettyTable
```

```python
nlp = spacy.load('en_core_web_sm')  # Loads the small English pre-trained model and its processing pipeline.
```

```python
text="Hello I'm learning NLP.This is the SpaCy lesson."
doc=nlp(text)
```

```python
doc
```

```python
doc.text
```

```python
doc.to_dict()
```

```python
for tocken in doc:
    print(tocken,tocken.is_alpha,tocken.is_punct)
    
```

```python
table=PrettyTable()
table.field_names=['token','is alpha','is punct','is number']
for token in doc:
    #print(token.text,token.is_alpha,token.is_punct,token.like_num)
    table.add_row([token.text,token.is_alpha,token.is_punct,token.like_num])
print(table)    
```
