### Regex-Regular expressions
There are different types of characters of a regex expression
- Metacharacters
- Quantifiers
- Groups and Ranges
- Escape Character or character classes


## 1.Matching a Pattern: re.match()

```python
import re
```

```python
text= """Hi!!,This is a NLP lesson.344564576
        65867596
        The mobile number is:7342985702.
        This is my email: learnnlpprof@gmail.com"""
```

```python
pattern='Hi!!,'
result=re.match(pattern=pattern,string=text) #For match it is necessary that your string has to start with the desired pattern.Here you cannot match the string present in between your main string.

result
```

```python
result.span()
```

```python
result.group()
```

```python
pattern=r'This' # there is no change in output with or without that r in the beginning.
result=re.match(pattern=pattern,string=text)
print(result)
```

## 2.Searching for a Pattern: re.search()

```python
print(text)
```

```python
pattern='This'
result=re.search(pattern=pattern,string=text)
result
```

```python
pattern='\d'
result=re.search(pattern=pattern,string=text)
result #output is just 3 as \d just matches the first digit
```

```python
pattern='\d+'# this will return the digit followed by any number of digits.
result=re.search(pattern=pattern,string=text)
result
```

```python
pattern=r'\w+@\w+.\w+' #Better write r, that indicates that you are writing this pattern for regex module. this is the simplest email pattern
result=re.search(pattern=pattern,string=text)
result
```

## 3.Finding All Matches: re.findall()

```python
print(text)
```

```python
pattern=r'\w+@\w+.\w+' #Better write r, that indicates that you are writing this pattern for regex module. this is the simplest email pattern
result=re.findall(pattern=pattern,string=text) #Slower than re.search()
result
```

```python
pattern=r'\d+' #Better write r, that indicates that you are writing this pattern for regex module. this is the simplest email pattern
result=re.findall(pattern=pattern,string=text) #Slower than re.search()
result
```

```python
pattern=r'\d{4}-\d{2}-\d{2}' 
result=re.findall(pattern=pattern,string="Today's date is 2030-05-17")
print(result)
```

## 4.Splitting a String: re.split()

```python
print(text)
```

```python
pattern=r'\d{4}-\d{2}-\d{2}' 
result=re.split(pattern=pattern,string="Today's date is 2030-05-17")
print(result)
```

```python
pattern=r'\d+' 
result=re.split(pattern=pattern,string=text) 
result
```

```python
len(result)
```

## 5.Replacing Matches: re.sub()

```python
print(text)
```

```python
pattern=r'0?\d{10}'
re.findall(pattern,text)
```

```python
pattern=r'0?\d{10}'
result=re.sub(pattern,"*",text)
print(result)
```

```python
print(text)
```

```python
def replace_with_star(match):
    return'*'*len(match.group())

pattern=r'0?\d{10}'
result=re.sub(pattern,replace_with_star,text)
print(result)
```

```python
def replace_with_star(match):
    return'*'*len(match.group())

pattern=r'0?\d{10}|\w+@\w+.\w+'
result=re.sub(pattern,replace_with_star,text)
print(result)
```
