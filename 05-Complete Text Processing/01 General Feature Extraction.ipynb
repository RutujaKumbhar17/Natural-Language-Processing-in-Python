### 2.Character Count

```python
text='This is character count lesson'
len(text.replace(' ','')) # need to do it as the spaces are not characters
```

```python
pattern=r'\s'
re.sub(pattern,'',text) #Does the same thing as above
```

```python
df['char_count']=df['tweets'].apply(lambda x: len(re.sub(pattern,'',x)))
#.apply(lambda x: ...): This is a "loop" alternative. It tells Pandas to take every row in the tweets column, one by one, and pass it (as variable x) into the function that follows.
```

### 3.Word Count

```python
x= 'word count example'
print(x.split())
len(x.split())
```

```python
df['word_count']=df['tweets'].apply(lambda x: len(x.split()))
```

```python
df.sample(5)
```

### 4. Average Word Length

```python
df['avg_word_len']= df['char_count']/df['word_count']
```

```python
df
```

```python
df['avg_word_len']=df['avg_word_len'].apply(lambda x: round(x, 2))
```

```python
df
```

### 5. Stop Words Count

```python
from spacy.lang.en.stop_words import STOP_WORDS as stopwords
```

```python
len(stopwords)
```

```python
x = "This is an example text data. let's use this"
([word for word in x.lower().split() if word in stopwords])
```

```python
x = "This is an example text data. let's use this"
len([word for word in x.lower().split() if word in stopwords])
```

```python
df['stop_words_len']=df['tweets'].apply(lambda x:len([word for word in x.lower().split() if word in stopwords]))
```

### 6. Count #HashTags and @Mentions 

```python
x='this is #cool that we can spot @mentions and #hashtags here'
```

```python
len(re.findall(r'#\w+',x)),len(re.findall(r'@\w+',x))
```

```python
df['hashtags_count']=df['tweets'].apply(lambda x:len(re.findall(r'#\w+',x)))
df['mensions_count']=df['tweets'].apply(lambda x:len(re.findall(r'@\w+',x)))
```

```python
df.sample(5)
```

```python
df['hashtags_count'].value_counts()
```

```python
df[df['hashtags_count']>0]
```

### 7.Numeric Digit Count

```python
x='This is the sample digit data 3 4 5 6 545 6564 '
print(re.findall(r'\d',x))
re.findall(r'\d+',x)
```

```python
#the above pattern is not right for the texts like below text.
x = 'I want coupon code for product KGP2312AU43 and KGP1243RT. I need 10 pcs.'
re.findall(r'\b\d+\b', x)
```

```python
df['numeric_count'] = df['tweets'].apply(lambda x: len(re.findall(r'\b\d+\b', x)))
```

```python
df[df['numeric_count']>0].sample(5)
```

### 8.UPPER CASE word Count

```python
x='UPPER CASE TEST is going on'
```

```python
[word for word in x.split() if word.isupper()], len([word for word in x.split() if word.isupper()])
```

```python
df['upper_counts'] = df['tweets'].apply(lambda x: len([word for word in x.split() if word.isupper()]))
```

```python
df[df['upper_counts']>0].sample(5)
```

```python
df[df['upper_counts']>3].iloc[0]['tweets'] #.iloc[0] (The Selector): Even if there are 500 tweets that meet your criteria, iloc[0] tells Pandas to grab only the first one in that filtered list.
```
