### 9. Lower Case Conversion 

```python
import pandas as pd
import numpy as np
import re
```

```python
url = 'https://github.com/laxmimerit/All-CSV-ML-Data-Files-Download/raw/master/twitter4000.csv'
df = pd.read_csv(url)
```

```python
x = 'IS bored and wants to watch a movie        any suggestions '
x.lower()

re.sub(r'\s+', ' ', x).lower()
```

```python
# we will make the changes on original column itself as we are preprocessing the data.
df['tweets']=df['tweets'].apply(lambda x: re.sub('\s+',' ',x).lower)
```

### 10. Contraction to Expansion 

```python
import json

contractions = json.load(open('data/contractions.json'))
```

```python
contractions.get("don't")
```

```python
# Dictionary lookup: .get(key, default) returns the expanded word if it exists,
# otherwise it keeps the original word to prevent errors.
x = "don't mess with me. I'm done with this stuff."
[contractions.get(word, word) for word in x.split()]
```

```python
[contractions.get(word.lower(), word) for word in x.split()]
```

```python
" ".join([contractions.get(word.lower(), word) for word in x.split()])
```

```python
df['tweets']=df['tweets'].apply(lambda x: " ".join([contractions.get(word.lower(), word) for word in str(x).split()])) #str(x)is imp as without str python is thinking that it's a method and not string.
```

### 11. Count and Remove Emails 

```python
x = 'contact me at rutujakumbhar@gmail.com and info@rutuja@gmail.com'

pattern = r'\w+@\w+.\w+'
result = re.findall(pattern=pattern, string=x)
print(result)
```

```python
x = 'contact me at rutujakumbhar@gmail.com and info@rutuja@gmail.co.in'

pattern = r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Z|a-z]{2,}\b'
result = re.findall(pattern=pattern, string=x)
print(result)
```

```python
df['emails'] = df['emails'] =df['tweets'].apply(lambda x: ",".join(re.findall(pattern=pattern, string=x)))
```

### 12. Count and Remove URLs

```python
x='Check this link: https://rutujakumbhar.com and www.rutujakumbhar.com'
pattern=r"http\S+|www\.\S+"
re.findall(pattern,x)
```

```python
df['urls']=df['tweets'].apply(lambda x: re.findall(pattern,x))
```
